```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
A = pd.read_csv("../data/temp/A_new.csv", index_col=0)

plt.figure(figsize=(12, 6)) 
sns.heatmap(A,cmap='Greys')
plt.xlabel('sub species')
plt.savefig('../data/temp/A.png', dpi=300, bbox_inches='tight') 
plt.show()
```

```python
Aorg = pd.read_csv("../data/temp/A_organics.csv", index_col=0)
```

```python
Aorg = Aorg / Aorg.max()
```

```python
plt.figure(figsize=(9.5, 6)) 
sns.heatmap(Aorg,cmap='Greys')
plt.xlabel('sub species')
plt.ylabel('species')
plt.savefig('../data/temp/Aorg.png', dpi=300, bbox_inches='tight') 
plt.show()
```

```python
V = pd.read_csv("../data/temp/V_aligned.csv",index_col=0)
```

```python
plt.figure(figsize=(12, 6))
for i in range(V.shape[1]):
    plt.plot(V.iloc[:,i],label=V.columns[i])
    
plt.xlabel('Wavenumber (cm⁻¹)')
plt.ylabel('MAC')
plt.legend()
plt.savefig('../data/temp/Vinit.png', dpi=300, bbox_inches='tight') 
plt.show()
```
